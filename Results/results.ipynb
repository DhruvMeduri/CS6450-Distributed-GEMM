```python
import pandas as pd
import matplotlib.pyplot as plt

# ---- Load CSV ----
df = pd.read_csv("timings_matrixsize_N4.csv")   

df = df.sort_values("matrix_size")

l1 = "A,B in shared"
l2 = "B in disc"
l3 = "B in shared"

# ---- Extract columns ----
x = df["matrix_size"]
y1 = df["compute"]
y2 = df["discB_compute"]
y3 = df["sharedB_compute"]

# ---- Plot ----
plt.figure(figsize=(10, 6))
plt.plot(x, y1, marker='o', label=l1)
plt.plot(x, y2, marker='s', label=l2)
plt.plot(x, y3, marker='^', label=l3)

plt.xlabel("Matrix Size")
plt.ylabel("Time (seconds)")
plt.title("Compute Methods vs Matrix Size with Number of Processes = 4")
plt.legend()
# plt.xticks(x)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
